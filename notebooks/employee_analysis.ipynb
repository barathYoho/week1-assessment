# Employee Data Analysis

## Project Overview
Analyze employee data using Python, NumPy and Pandas.

```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv('../data/employees.csv')
print('Dataset loaded successfully')
display(df.head())
```

## 1. Dataset Information

```
print('Dataset Shape:', df.shape)
df.info()
display(df.describe())
```

## 2. Salary Analysis

```
print('Average Salary:', df['salary'].mean())
print('Median Salary:', df['salary'].median())
print('Minimum Salary:', df['salary'].min())
print('Maximum Salary:', df['salary'].max())
```

## 3. Department-wise Analysis

```
department_summary = df.groupby('department').agg(headcount=('id','count'), average_salary=('salary','mean'), maximum_salary=('salary','max'))
display(department_summary)
```

## 4. Chennai Employees

```
chennai_employees = df[df['city'] == 'Chennai']
display(chennai_employees)
print('Chennai Employee Count:', len(chennai_employees))
```

## 5. Performance Measurement

```
%timeit df.groupby('department')['salary'].mean()
```

## 6. Key Findings

1. The dataset contains 8 employee records.
2. IT has 3 employees.
3. Finance has the highest average salary of Rs. 90,000.
4. Sales and HR have 2 employees each.
5. Four employees belong to Chennai.
